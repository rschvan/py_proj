## Proximity, PFnet, Collection, Netpic classes demo

```python
from pypf import Proximity, PFnet, Collection, Netpic
import matplotlib.pyplot as plt

bioprx = Proximity("pypf/data/statecaps.prx.xlsx")
print('bioprx: ', bioprx.name, ' ', bioprx.termsid)
bionet = PFnet(bioprx)
print(f"{bionet.name} {bionet.nlinks} links")
psyprx = Proximity("pypf/data/psy.prx.xlsx")
print('psyprx: ', psyprx.name, ' ', psyprx.termsid)
psynet = PFnet(psyprx)
print(f"{psynet.name} {psynet.nlinks} links")
col = Collection()
col.add_proximity(bioprx)
col.add_proximity(psyprx)
col.add_pfnet(bionet)
col.add_pfnet(psynet)
print(col.proximities.keys())
print(col.pfnets.keys())
biopic = Netpic(bionet)
biopic.create_view()

#print(f"biopic = {biopic.fig}")
#print(f" biopic {biopic.net.coords[0,]}")
# psypic = Netpic(psynet)
# print(f" pic {psypic.net.name}")
```
